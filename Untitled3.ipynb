```python
from sklearn.model_selection import train_test_split

y = data['labels']

columns_to_exclude_from_X = [
    'labels', 'status', 'Unnamed: 0', 'id', 'object_id', 'name',
    'zip_code', 'state_code', 'state_code.1', 'Unnamed: 6', 'city',
    'founded_at', 'closed_at', 'first_funding_at', 'last_funding_at',
    'State_code', 'State'
]

existing_columns_to_exclude = [col for col in columns_to_exclude_from_X if col in data.columns]

X = data.drop(existing_columns_to_exclude, axis=1)

# Handle categorical 'category' column if it exists and wasn't dropped in earlier steps
if 'category' in X.columns:
    X = pd.get_dummies(X, columns=['category'], prefix='category', drop_first=False)

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size =0.3, random_state = 0)
```

```python
rf = RandomForestClassifier()
param_grid ={'n_estimators':[100,200,300],
             'max_depth':[10,20,30],
             'min_samples_split':[2,5,6],
             'min_samples_leaf':[1,2,3],
             'bootstrap':[True, False]}
grid_search = GridSearchCV(estimator=rf, param_grid=param_grid, cv= 5, n_jobs= -1, verbose = False)
grid_search.fit(x_train, y_train)
print('Best parameters:', grid_search.best_params_)
```

```python
from flask import Flask, render_template, request
import joblib
import pandas as pd

# Initialize the Flask application only if it doesn't exist globally
if 'app' not in globals():
    app = Flask(__name__)

    # Load the trained model
    model = joblib.load('random_forest_model.pkl')

    # Define the features that your model expects, in the correct order.
    # This list should match the columns of X used during model training.
    # Based on x_train.columns.tolist() output from earlier
    MODEL_FEATURES = [
        'latitude', 'longitude', 'age_first_funding_year', 'age_last_funding_year',
        'age_first_milestone_year', 'age_last_milestone_year', 'relationships',
        'funding_rounds', 'funding_total_usd', 'milestones', 'is_CA', 'is_NY', 'is_MA',
        'is_TX', 'is_otherstate', 'has_VC', 'has_angel', 'has_roundA', 'has_roundB',
        'has_roundC', 'has_roundD', 'avg_participants', 'is_top500', 'founded_year',
        'category_biotech', 'category_enterprise', 'category_games_video',
        'category_hardware', 'category_mobile', 'category_network_hosting',
        'category_others', 'category_semiconductor', 'category_software', 'category_web'
    ]

    # Define the home route
    @app.route('/')
    def home():
        return render_template('index.html')

    # Define the predict route
    @app.route('/predict', methods=['POST'])
    def predict():
        # Get the input values from the form
        age_first_funding_year = float(request.form['age_first_funding_year'])
        age_last_funding_year = float(request.form['age_last_funding_year'])
        age_first_milestone_year = float(request.form['age_first_milestone_year'])
        age_last_milestone_year = float(request.form['age_last_milestone_year'])
        relationships = float(request.form['relationships'])
        funding_rounds = float(request.form['funding_rounds'])
        funding_total_usd = float(request.form['funding_total_usd'])
        milestones = float(request.form['milestones'])
        avg_participants = float(request.form['avg_participants'])

        # Create a list with the input values
        input_data = [
            age_first_funding_year,
            age_last_funding_year,
            age_first_milestone_year,
            age_last_milestone_year,
            relationships,
            funding_rounds,
            funding_total_usd,
            milestones,
            avg_participants
        ]

        # Make a prediction using the loaded model
        prediction = model.predict([input_data])[0]

        # Map the predicted label to a meaningful output
        if prediction == 1:
            result = 'Acquired'
        else:
            result = 'Closed'

        # Render the prediction result
        return render_template('result.html', result=result)
else:
    print("Flask app already initialized. Skipping re-initialization and route registration.")
```

```python
import threading
import time
import socket

# Kill any existing ngrok tunnels to ensure a clean slate
ngrok.kill()
print("Killed all active ngrok tunnels.")

# Function to check if a port is truly free by attempting to bind to it
def is_port_free(port):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        try:
            # Try to bind to the port. If successful, the port is free.
            s.bind(("127.0.0.1", port))
            return True
        except OSError:
            # If binding fails, the port is in use.
            return False

flask_port = 5000

# Define the function to run the Flask application
def run_flask_app_thread():
    global app # Ensure 'app' is accessible from the global scope
    print(f"Attempting to start Flask app on port {flask_port} in new thread.")
    try:
        # `app.run()` is a blocking call, it will keep the thread alive.
        app.run(port=flask_port, debug=False, use_reloader=False)
    except OSError as e:
        print(f"Flask app failed to bind to port {flask_port}: {e}")
    except Exception as e:
        print(f"Flask app thread encountered an unexpected error: {e}")

# --- Main execution logic ---

if not is_port_free(flask_port):
    print(f"Port {flask_port} is currently in use. A previous Flask instance might be active.")
    print("Please restart the Colab runtime and try again if you want to start a fresh Flask app.")
    print(f"Attempting to establish ngrok tunnel to the *existing* listener on port {flask_port} (if any).")
    try:
        # If the port is in use, we try to connect ngrok to whatever is already there.
        # This could be a previous working Flask instance or a zombie process.
        public_url = ngrok.connect(flask_port)
        print(f"ngrok tunnel established at: {public_url}")
        print("\nTo interact with the API, send a POST request to the /predict endpoint of the public URL with your form data.")
    except Exception as e:
        print(f"Failed to establish ngrok tunnel: {e}")
else:
    # Port is free, proceed to start a new Flask app
    print("Starting new Flask app thread.")
    flask_thread = threading.Thread(target=run_flask_app_thread, name='FlaskAppThread', daemon=True)
    flask_thread.start()

    # Wait for the Flask app to actually listen on the port
    start_time = time.time()
    timeout = 20 # seconds to wait for Flask to bind
    # Loop while the port is still free (meaning our Flask app hasn't bound yet)
    while is_port_free(flask_port) and (time.time() - start_time) < timeout:
        time.sleep(0.5)

    if not is_port_free(flask_port): # Port is now in use, hopefully by our Flask app
        print(f"Flask app is successfully listening on port {flask_port}.")
        try:
            public_url = ngrok.connect(flask_port)
            print(f"ngrok tunnel established at: {public_url}")
            print("\nTo interact with the API, send a POST request to the /predict endpoint of the public URL with your form data.")
        except Exception as e:
            print(f"Failed to establish ngrok tunnel: {e}")
    else:
        print(f"Flask app did not start listening on port {flask_port} within {timeout} seconds. Please check logs for errors or restart runtime.")
```

```python
%%writefile app.py
import streamlit as st
import requests

st.title("🚀 Startup Status Predictor")
st.markdown("Enter the startup details below to predict if it will be 'Acquired' or 'Closed'.")

# Input fields for the Streamlit app
# These should match the features your Flask API expects
col1, col2 = st.columns(2)

with col1:
    latitude = st.number_input("Latitude", value=37.77)
    longitude = st.number_input("Longitude", value=-122.41)
    founded_year = st.number_input("Founded Year", value=2010, min_value=1900, max_value=2026)
    age_first_funding_year = st.number_input("Age First Funding Year", value=0.5)
    age_last_funding_year = st.number_input("Age Last Funding Year", value=2.0)
    age_first_milestone_year = st.number_input("Age First Milestone Year", value=1.0)
    age_last_milestone_year = st.number_input("Age Last Milestone Year", value=3.0)
    relationships = st.number_input("Relationships", value=5, min_value=0)
    funding_rounds = st.number_input("Funding Rounds", value=2, min_value=1)

with col2:
    funding_total_usd = st.number_input("Funding Total (USD)", value=5000000)
    milestones = st.number_input("Milestones", value=2, min_value=0)
    avg_participants = st.number_input("Average Participants", value=2.5)

    # State and Category booleans/dummies
    is_CA = st.checkbox("Is in CA", value=True)
    is_NY = st.checkbox("Is in NY")
    is_MA = st.checkbox("Is in MA")
    is_TX = st.checkbox("Is in TX")
    is_otherstate = st.checkbox("Is in Other State")

    has_VC = st.checkbox("Has VC Funding", value=True)
    has_angel = st.checkbox("Has Angel Funding")
    has_roundA = st.checkbox("Has Round A Funding", value=True)
    has_roundB = st.checkbox("Has Round B Funding")
    has_roundC = st.checkbox("Has Round C Funding")
    has_roundD = st.checkbox("Has Round D Funding")
    is_top500 = st.checkbox("Is Top 500 Startup", value=False)

    category = st.selectbox("Category", [
        "biotech", "enterprise", "games_video", "hardware", "mobile",
        "network_hosting", "others", "semiconductor", "software", "web"
    ], index=8) # Default to 'software'


# Button to trigger prediction
if st.button("Predict Status"):
    # Define your ngrok URL here. This should be the one from cell a0700ce0's output.
    # IMPORTANT: Update this URL if your ngrok tunnel changes!
    flask_api_url = "https://unjarred-rae-glaucously.ngrok-free.dev/predict" # Placeholder, replace with actual ngrok URL

    # Construct the payload for the Flask API
    # Ensure feature order and names exactly match MODEL_FEATURES
    payload = {
        "latitude": latitude,
        "longitude": longitude,
        "founded_year": founded_year,
        "age_first_funding_year": age_first_funding_year,
        "age_last_funding_year": age_last_funding_year,
        "age_first_milestone_year": age_first_milestone_year,
        "age_last_milestone_year": age_last_milestone_year,
        "relationships": relationships,
        "funding_rounds": funding_rounds,
        "funding_total_usd": funding_total_usd,
        "milestones": milestones,
        "avg_participants": avg_participants,
        "is_CA": int(is_CA),
        "is_NY": int(is_NY),
        "is_MA": int(is_MA),
        "is_TX": int(is_TX),
        "is_otherstate": int(is_otherstate),
        "has_VC": int(has_VC),
        "has_angel": int(has_angel),
        "has_roundA": int(has_roundA),
        "has_roundB": int(has_roundB),
        "has_roundC": int(has_roundC),
        "has_roundD": int(has_roundD),
        "is_top500": int(is_top500),
        "category_biotech": 0, # Default to 0, will be 1 if selected
        "category_enterprise": 0,
        "category_games_video": 0,
        "category_hardware": 0,
        "category_mobile": 0,
        "category_network_hosting": 0,
        "category_others": 0,
        "category_semiconductor": 0,
        "category_software": 0,
        "category_web": 0
    }
    # Set the selected category to 1
    payload[f"category_{category}"] = 1

    try:
        response = requests.post(flask_api_url, json=payload)
        if response.status_code == 200:
            prediction_result = response.json().get('prediction')
            if prediction_result == "Closed":
                st.error(f"The model predicts: {prediction_result}")
            else:
                st.success(f"The model predicts: {prediction_result}")
        else:
            st.error(f"Error from API: Status Code {response.status_code} - {response.text}")
    except requests.exceptions.ConnectionError as e:
        st.error(f"Connection Error: Could not connect to the Flask API. Is it running? ({e})")
    except Exception as e:
        st.error(f"An unexpected error occurred: {e}")
```

```python
import subprocess
import time
import socket

# Force uninstall and then install pyngrok to ensure a fresh install in the current environment
print("Attempting to uninstall pyngrok...")
subprocess.run(['pip', 'uninstall', '-y', 'pyngrok'], capture_output=True, text=True)
print("Attempting to install pyngrok...")
subprocess.run(['pip', 'install', 'pyngrok'], capture_output=True, text=True, check=True)
print("pyngrok installation attempt completed.")

from pyngrok import ngrok # Import after fresh installation

# 1. Kill any existing Streamlit processes (if running)
subprocess.run(['killall', 'streamlit'], capture_output=True, text=True, check=False)
print("Killed existing Streamlit processes (if any).")

# 2. Kill any existing ngrok tunnels
ngrok.kill()
print("Killed all active ngrok tunnels.")

# Define the Streamlit port
streamlit_port = 8501

# Function to check if a port is truly free
def is_port_free(port):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        try:
            s.bind(("127.0.0.1", port))
            return True
        except OSError:
            return False

# 3. Launch Streamlit in the background
streamlit_command = f"nohup streamlit run app.py --server.port {streamlit_port} > streamlit_output.log 2>&1"
subprocess.Popen(streamlit_command, shell=True)
print(f"Streamlit launched in background on port {streamlit_port}. Waiting for it to start...")
time.sleep(10) # Give Streamlit ample time to start

# Wait for the Streamlit app to actually listen on the port
start_time = time.time()
timeout = 20 # seconds to wait for Streamlit to bind
while is_port_free(streamlit_port) and (time.time() - start_time) < timeout:
    time.sleep(0.5)

if not is_port_free(streamlit_port):
    print(f"Streamlit app is successfully listening on port {streamlit_port}.")
    # 4. Establish an ngrok tunnel
    try:
        # Ensure ngrok authtoken is set (already done in previous cells, but good to be explicit if this were a standalone script)
        # ngrok.set_auth_token("YOUR_AUTHTOKEN") # Replace with actual authtoken if not set globally

        public_url = ngrok.connect(streamlit_port)
        print(f"\nPublic Streamlit URL: {public_url}")
        print(f"Internal Streamlit URL: http://localhost:{streamlit_port}")
        print("Check 'streamlit_output.log' for Streamlit's logs.")
    except Exception as e:
        print(f"Failed to establish ngrok tunnel: {e}")
else:
    print(f"Streamlit app did not start listening on port {streamlit_port} within {timeout} seconds. Please check 'streamlit_output.log' for errors or restart runtime.")
```
